```python
import fastf1
fastf1.Cache.enable_cache('../data')  # points to the shared data folder
```

```python
import pandas as pd
laps = pd.read_csv('../shared/cleaned_laps_2023_Qatar.csv')
laps
```

```python
session = fastf1.get_session(2023, 'Qatar Grand Prix', 'R')
session.load()
```

```python
all_laps = session.laps
all_laps
```

```python
pit_in_out_laps = all_laps[(all_laps['PitInTime'].notna()) | (all_laps['PitOutTime'].notna())]
pit_in_out_laps = pit_in_out_laps.reset_index(drop = True)
pit_in_out_laps
```

```python
# shift PitOutTime to the same lap as PitInTime
pit_in_out_laps['PitOutTime(updated)'] = pit_in_out_laps.groupby('Driver')['PitOutTime'].shift(-1)
pit_in_out_laps
```

There can be few missing values for PitOutTime(updated) which can be due to the DNF during the pit stop or the race ended on their final pit stop.

There can be few missing values for PitInTime which can be due to starting from pitlane.

```python
pit_in_out_laps.head(50)
```

```python
pit_in_out_laps.tail(60)
```

PER on lap 1, PitOutTime is given but no PitInTime(no lap is before lap 1) because he started from pitlane due to change in additional power-unit elements after his sprint crash.

https://www.formula1.com/en/latest/article/official-grid-verstappen-on-pole-perez-takes-pit-lane-start-after-power-unit.4LWtIkRIcXDAQUOhezgtUY

SAR on lap 40 has a PitInTime but no PitOutTime as retired on lap 40.

https://www.formula1.com/en/results/2023/races/1221/qatar/race-result

```python
pit_stop = pit_in_out_laps[pit_in_out_laps['PitInTime'].notna()].copy() # new df with only pit stops
pit_stop
```

```python
pit_stop['PitInTime(seconds)'] = pit_stop['PitInTime'].dt.total_seconds()
pit_stop['PitOutTime(seconds)'] = pit_stop['PitOutTime(updated)'].dt.total_seconds()
pit_stop
```

```python
pit_stop.columns
```

```python
pit_stop['PitStopDuration'] = pit_stop['PitOutTime(seconds)'] - pit_stop['PitInTime(seconds)']
pit_stop = pit_stop[['Driver', 'Team', 'LapNumber', 'PitInTime(seconds)', 'PitOutTime(seconds)', 'PitStopDuration']]
pit_stop
```

SAR retired on lap 40 after going into the pits.

```python
# retirements
retirements = pit_stop[pit_stop['PitStopDuration'].isna()].copy()
retirements['Status'] = 'Retired in pit' # if PitStopDuration = NaN, either retired in pits or took a pit stop at last lap
retirements
```

```python
complete_pit_stop = pit_stop[pit_stop['PitStopDuration'].notna()].copy()
complete_pit_stop
```

```python
team_avg_pit = complete_pit_stop.groupby('Team')['PitStopDuration'].mean().sort_values()
team_avg_pit
```

```python
# flagging outliers
mean_pit = complete_pit_stop['PitStopDuration'].mean()
std_pit = complete_pit_stop['PitStopDuration'].std()
threshold = mean_pit + (2 * std_pit)
print(threshold)
outliers = complete_pit_stop[complete_pit_stop['PitStopDuration'] > threshold].copy()
outliers
```

```python
normal_pit_stop = complete_pit_stop[complete_pit_stop['PitStopDuration'] <= threshold]
normal_pit_stop
```

```python
team_avg_pit = normal_pit_stop.groupby('Team')['PitStopDuration'].mean().sort_values()
team_avg_pit
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize = (12, 6))
sns.barplot(
    x = team_avg_pit.values,
    y = team_avg_pit.index
)
plt.xlim(team_avg_pit.min() - 1, team_avg_pit.max() + 1)
plt.xlabel('Average Pit Stop Duration(seconds)')
plt.title('Average Pit Stop Time by Team(Fastest to Slowest)')
plt.savefig('charts/avg_pit_stop_by_team.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize = (12, 8))
sns.boxplot(
    data = normal_pit_stop,
    x = 'Team',
    y = 'PitStopDuration'
)
plt.title('Pit Stop Consistency by Team')
plt.savefig('charts/pit_stop_consistency_by_team.png', dpi=300, bbox_inches='tight')
plt.show()
```

A team could have fast average pit stop but highly inconsistent.

```python
# fastest single pit stop per team
fastest_per_team = normal_pit_stop.groupby('Team')['PitStopDuration'].min().sort_values()
fastest_per_team
```

```python
plt.figure(figsize = (12, 6))
sns.barplot(
    x = fastest_per_team.values,
    y = fastest_per_team.index
)
plt.xlim(fastest_per_team.min() - 1, fastest_per_team.max() + 1)
plt.xlabel('Fastest Pit Stop(seconds)')
plt.title('Fastest Single Pit Stop By Team')
plt.savefig('charts/fastest_single_pit_stop_by_team.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
plt.figure(figsize = (10, 4))
sns.scatterplot(
    data = normal_pit_stop,
    x = 'LapNumber',
    y = 'PitStopDuration',
    hue = 'Team'
)
plt.title('Pit Stop Duration Over the Race')
plt.legend(
    bbox_to_anchor = (1.02, 1)
)
plt.tight_layout()
plt.savefig('charts/pit_stop_duration_over_the_race.png', dpi=300, bbox_inches='tight')
plt.show()
```

Pit stop went faster or slower as the race progressed.

```python
# calculating mean and std to be used in Monte Carlo Simulation
mean_pit_duration = normal_pit_stop['PitStopDuration'].mean()
std_pit_duration = normal_pit_stop['PitStopDuration'].std()
# returns a numpy.float so convert into python float
pit_stats = {
    'mean pit duration' : float(mean_pit_duration),
    'std pit duration' : float(std_pit_duration)
}
pit_stats
```

```python
# save as json to get just the values
import json
with open('../shared/pit_stats.json', 'w') as f:
    json.dump(pit_stats, f)
```
