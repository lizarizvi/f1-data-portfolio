```python
import fastf1
fastf1.Cache.enable_cache('../data')  # points to the shared data folder
```

```python
import pandas as pd
laps = pd.read_csv('../shared/cleaned_laps_2023_Qatar.csv')
laps
```

```python
session = fastf1.get_session(2023, 'Qatar Grand Prix', 'R')
session.load()
```

```python
all_laps = session.laps
all_laps
```

```python
pit_in_out_laps = all_laps[(all_laps['PitInTime'].notna()) | (all_laps['PitOutTime'].notna())]
pit_in_out_laps = pit_in_out_laps.reset_index(drop = True)
pit_in_out_laps
```

```python
# shift PitOutTime to the same lap as PitInTime
pit_in_out_laps['PitOutTime(updated)'] = pit_in_out_laps.groupby('Driver')['PitOutTime'].shift(-1)
pit_in_out_laps
```

There can be few missing values for PitOutTime(updated) which can be due to the DNF during the pit stop or the race ended on their final pit stop.

There can be few missing values for PitInTime which can be due to starting from pitlane.

```python
pit_stop = pit_in_out_laps[pit_in_out_laps['PitInTime'].notna()].copy() # new df with only pit stops
pit_stop
```

```python
pit_stop['PitInTime(seconds)'] = pit_stop['PitInTime'].dt.total_seconds()
pit_stop['PitOutTime(seconds)'] = pit_stop['PitOutTime(updated)'].dt.total_seconds()
pit_stop
```

```python
pit_stop.columns
```

```python
pit_stop['PitStopDuration'] = pit_stop['PitOutTime(seconds)'] - pit_stop['PitInTime(seconds)']
pit_stop = pit_stop[['Driver', 'Team', 'LapNumber', 'PitInTime(seconds)', 'PitOutTime(seconds)', 'PitStopDuration']]
pit_stop
```

SAR retired on lap 40 after going into the pits.

```python
# retirements
retirements = pit_stop[pit_stop['PitStopDuration'].isna()].copy()
retirements['Status'] = 'Retired in pit' # if PitStopDuration = NaN, either retired in pits or took a pit stop at last lap
retirements
```

```python
complete_pit_stop = pit_stop[pit_stop['PitStopDuration'].notna()].copy()
complete_pit_stop
```

```python
team_avg_pit = complete_pit_stop.groupby('Team')['PitStopDuration'].mean().sort_values()
team_avg_pit
```
