```python
import fastf1
fastf1.Cache.enable_cache('../data')  # points to the shared data folder
```

```python
session = fastf1.get_session(2023, 'Qatar Grand Prix', 'R')  # R = race
session.load()
```

```python
results = session.results
results
```

```python
# lap data
laps = session.laps
laps
```

```python
# clean data
laps = laps[laps['TrackStatus'] == '1'] # 1 == green flag
laps = laps[laps['PitInTime'].isna() & laps['PitOutTime'].isna()] # pit-in and pit-out laps are slow
laps = laps[laps['Deleted'] == False] # only valid laps(remove deleted laptimes)
laps
```

```python
# laptime from TimeDelta type to seconds
laps['LapTime(seconds)'] = laps['LapTime'].dt.total_seconds()
laps
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize = (10, 8))
sns.boxplot(
    data = laps,
    x = 'Driver',
    y = 'LapTime(seconds)'
)
plt.title('Lap Time Distribution by Driver')
plt.savefig('charts/lap_time_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
```

Box position shows the pace, box height shows the consistency.

```python
# consistency -> lower std, more consistent
consistency = laps.groupby('Driver')['LapTime(seconds)'].agg(['mean', 'std']).reset_index()
consistency.columns = ['Driver', 'AvgLapTime', 'Consistency(std)']
consistency
```

```python
laps.columns
```

```python
consistency = consistency.sort_values('Consistency(std)')
plt.figure(figsize = (10, 8))
sns.barplot(
    data = consistency,
    x = 'Consistency(std)',
    y = 'Driver'
)
plt.title('Driver Consistency')
plt.savefig('charts/driver_consistency.png', dpi=300, bbox_inches='tight')
plt.show()
```

Lower std = more consistent driver.

```python
top_5_drivers = results.sort_values('Position').head()['Abbreviation']
top_5_drivers
```

```python
top_5_lap = laps[laps['Driver'].isin(top_5_drivers)][['Driver', 'LapNumber', 'LapTime(seconds)']]
top_5_lap
```

```python
plt.figure(figsize = (12, 6))
sns.lineplot(
    data = top_5_lap,
    x = 'LapTime(seconds)',
    y = 'LapNumber',
    hue = 'Driver'
)
plt.title('Lap Time Evolution Throughout the Race')
plt.savefig('charts/top_5_lap_time_evolution.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# comparing sector times
laps['Sector1Time(seconds)'] = laps['Sector1Time'].dt.total_seconds()
laps['Sector2Time(seconds)'] = laps['Sector2Time'].dt.total_seconds()
laps['Sector3Time(seconds)'] = laps['Sector3Time'].dt.total_seconds()
laps
```

```python
sector_average = laps.groupby('Driver')[['Sector1Time(seconds)', 'Sector2Time(seconds)', 'Sector3Time(seconds)']].mean()
sector_average
```

```python
fastest_per_sector = sector_average.min()
fastest_per_sector
```

```python
sector_gap = sector_average - fastest_per_sector
sector_gap
```

```python
sector_gap.columns = ['Sector1Gap', 'Sector2Gap', 'Sector3Gap']
sector_gap
```

```python
plt.figure(figsize = (12, 6))
sns.heatmap(
    data = sector_gap,
    annot = True,
    cmap = 'RdYlGn_r',
    fmt = '.3f'
)
plt.title('Time Lost per Sector (seconds behind fastest)')
plt.savefig('charts/heatmap_time_lost_per_sector.png', dpi=300, bbox_inches='tight')
plt.show()
```

Green = closer to the fastest driver.

Red = losing time, far from the fastest driver.

```python
# fastest laps comparison
fastest_lap = laps.groupby('Driver')['LapTime(seconds)'].min().sort_values()
fastest_lap
```

```python
plt.figure(figsize = (12, 8))
sns.barplot(
    x = fastest_lap.values,
    y = fastest_lap.index
)
plt.xlim(fastest_lap.min() - 1, fastest_lap.max() + 1)
plt.xlabel('Fastest LapTime(seconds)')
plt.title('Fastest Lap per Driver')
plt.savefig('charts/fastest_laptimes.png', dpi=300, bbox_inches='tight')
plt.show()
```
